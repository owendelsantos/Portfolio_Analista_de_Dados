```python
api_base_url = 'https://api.github.com/'
owner = 'amzn'

url_amzn = f'{api_base_url}/users/{owner}/repos'
```

Autenticação

```python
acesso_token = os.getenv("GITHUB_TOKEN", "SEU_TOKEN_AQUI")
headers = {
    "Authorization": f"Bearer {acesso_token}",
    "Accept": "application/vnd.github.v3+json"
}
```

Paginando os repositórios

```python
repos_list = []
for page_num in range (1, 7):
  try:
    url_page = f'{url_amzn}?page={page_num}'
    response = requests.get(url_page, headers = headers)
    repos_list.append(response.json())
  except: repos_list.append(None)
```

```python
len(repos_list)
```

```python
len(repos_list[0])
```

# ***Transformação dos dados***

Nome dos repositórios

```python
repos_list[0][1]
```

```python
repos_list[0][1]['name']
```

```python
repos_name = []

for page_name in repos_list:
  for repo in page_name:
    repos_name.append(repo['name'])
```

```python
len(repos_name)
```

Linguagem utilizada nos repositórios

```python
repos_list[0][1]
```

```python
repos_list[0][1]['language']
```

```python
repos_language = []

for page_name in repos_list:
  for repo in page_name:
    repos_language.append(repo['language'])
```

```python
len(repos_language)
```

Criando um DataFrame com os dados tratados

```python
import pandas as pd
```

```python
dados_amazon = pd.DataFrame()
dados_amazon['Nome repositório'] = repos_name
dados_amazon['Linguagem do repositório'] = repos_language
```
